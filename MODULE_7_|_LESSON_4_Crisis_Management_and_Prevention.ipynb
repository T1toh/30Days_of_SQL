## **Introduction**

Financial crises are the black swans that everyone sees coming – until they arrive and surprise absolutely everyone. This paradox captures the essence of crisis management: we know crises will occur, yet we consistently fail to predict their timing, magnitude, or precise form. The 2008 financial crisis wasn't the first to catch experts off guard, and it certainly won't be the last.

More precisely, crisis management represents the systematic attempt to identify, prevent, and respond to systemic breakdowns in financial markets. It encompasses everything from central bank emergency lending to the sophisticated computational methods that scan terabytes of market data for early warning signals. But here's the catch (and it's a big one): most crisis management frameworks are calibrated on the previous crisis, not the next one.

It pays to examine why this field has become so crucial in modern finance. The interconnectedness of global markets means that what starts as a housing bubble in Cleveland can end up freezing credit markets in London. Traditional risk management, focused on individual institutions, proved woefully inadequate when the entire system began to unravel simultaneously. The challenge isn't just technical – it's fundamentally about understanding how complex systems fail.

## **1. Central Bank Intervention Approaches**

### **1.1 Traditional Intervention Mechanisms**

Central banks are the fire departments of the financial world – except they're fighting fires that can spontaneously reignite and spread through invisible channels. For decades, their primary tool was elegantly simple: adjust short-term interest rates to cool overheated markets or stimulate sluggish economies. Lower rates during bad times, raise them during good times. What could go wrong?

The answer, as 2008 demonstrated brutally, is everything. When crisis hits and rates are already near zero, traditional monetary policy becomes as useful as a fire hose without water. This is the zero lower bound problem, though calling it a "problem" understates its significance – it's more like discovering your emergency brake doesn't work while hurtling toward a cliff.

Beyond rate adjustments, central banks maintain lender-of-last-resort facilities. Think of these as emergency credit lines for banks facing temporary liquidity crunches. The Federal Reserve's discount window, the ECB's marginal lending facility – these mechanisms work beautifully when the problem is one troubled bank (why?). But what happens when the entire banking system needs emergency liquidity simultaneously? The scale of intervention required can dwarf anything in the central bank's historical experience.

### **1.2 Unconventional Monetary Policies**

When traditional tools fail, central banks become remarkably creative. Quantitative easing emerged as the marquee unconventional policy – a fancy name for creating money electronically and using it to buy bonds. Lots of bonds. The Federal Reserve's balance sheet expanded from under $1 trillion to over $4.5 trillion, an expansion that would have been unthinkable before 2008.

Forward guidance represents perhaps an even more radical innovation. Central banks explicitly promise to keep rates low for extended periods or until specific conditions are met. This is monetary policy by commitment rather than action – the power of words rather than deeds. But there's a catch: credibility matters enormously. Markets must believe the central bank will stick to its promises even when circumstances change (and they always do).

The ultimate unconventional policy? Negative interest rates. Yes, central banks now charge banks for holding excess reserves, turning centuries of economic intuition on its head. It's like paying someone to borrow your money – economically bizarre but surprisingly effective at forcing banks to lend rather than hoard liquidity.

## **2. Crisis Management Frameworks**

### **2.1 Early Warning Systems and Indicators**

Crisis management frameworks attempt to solve an impossible problem: predicting the unpredictable. Early warning systems monitor dozens of indicators, from credit-to-GDP ratios to obscure measures of market microstructure, searching for patterns that might signal trouble ahead. The challenge is distinguishing genuine warning signals from false alarms – after all, no one wants to trigger a crisis by warning about one (why not?).

CoVaR, which we explored in Module 1, represents a sophisticated attempt to capture systemic risk. Unlike traditional risk measures that focus on individual institutions, CoVaR measures how one institution's distress affects the entire system. It's the difference between asking "How bad could this bank's losses be?" and "How bad could the system's losses be if this bank fails?" The distinction matters enormously when cascading failures can amplify initial shocks.

Machine learning techniques increasingly enhance these systems, though with mixed results. Deep learning models can process vast amounts of unstructured data – news sentiment, satellite imagery, even Twitter feeds – to detect early stress signals. But here's the uncomfortable truth: most financial crises involve unprecedented combinations of factors that no historical model can anticipate.

### **2.2 Crisis Response Protocols**

When crisis strikes, frameworks matter less than execution speed. Modern crisis management protocols establish clear decision-making hierarchies, communication channels, and escalation procedures. The Financial Stability Board coordinates international responses, ensuring that emergency actions in one country don't inadvertently destabilize others (though this coordination often breaks down precisely when it's needed most).

Crisis response typically follows a predictable sequence. First comes liquidity provision – central banks flood markets with cash to prevent funding spirals. Then comes the triage phase: which institutions can be saved, which must be allowed to fail, and which require special resolution procedures? This is where theory meets brutal reality, as political pressures and systemic concerns often override carefully planned protocols.

The resolution phase presents the greatest challenges. "Living wills" and bail-in mechanisms are supposed to provide orderly ways to wind down failing institutions without taxpayer bailouts. In practice, however, regulators often discover that systemically important institutions are simply too interconnected to fail cleanly. The result? Ad hoc interventions that bear little resemblance to the elegant frameworks designed during peacetime.

## **3. Post-Crisis Model Recalibration Using Robust Statistics**

### **3.1 Identifying Model Failures**

Financial crises are like stress tests that no one volunteered to take. They reveal every flaw, every hidden assumption, every parameter that worked perfectly until it didn't. Post-crisis forensic analysis typically uncovers a familiar pattern: models calibrated during calm periods systematically underestimate risks during turbulent ones (and why might this be inevitable rather than just unfortunate?).

The most common failure? Correlation assumptions that prove spectacularly wrong when markets crash. Diversification benefits disappear precisely when investors need them most, as assets that seemed uncorrelated suddenly move in lockstep. It's like discovering your umbrella is made of sugar just as the storm arrives.

Parameter instability represents another pervasive problem. Risk models often assume that statistical relationships remain constant over time, but crises frequently involve regime changes that render historical calibrations meaningless. The challenge isn't just technical – it's philosophical. How do you model events that have never happened before using data that excludes similar events?

### **3.2 Robust Statistical Methods**

Robust statistics offer partial solutions to these challenges by designing methods that perform reasonably well even when standard assumptions fail. Think of robust estimators as the all-terrain vehicles of statistical analysis – they may not win races on smooth highways, but they keep running when the road gets rough.

Consider the median absolute deviation (MAD) versus standard deviation. Standard deviation squares deviations, giving enormous weight to outliers – exactly what you don't want during crises when outliers contain the most important information. MAD, by contrast, maintains reasonable behavior even when data includes extreme observations. It's like having a risk measure that doesn't panic when markets do.

Robust covariance estimation using techniques like the Minimum Covariance Determinant (MCD) addresses the correlation breakdown problem directly. These methods identify the core relationships in data while downweighting observations that don't fit the pattern. But here's the subtlety: in finance, those "outlying" observations often represent the most valuable information about tail risks.

Regime-switching models acknowledge that markets operate under different statistical rules during normal and crisis periods. Rather than pretending these differences don't exist, regime-switching models explicitly incorporate multiple states with different parameters. It's like having different speed limits for different weather conditions – the same road requires different rules when visibility drops.

## **4. Structural Weaknesses Identification**

### **4.1 Leverage and Interconnectedness**

Excessive leverage is the original sin of finance – it amplifies profits during good times and losses during bad times, with predictable results. Every major financial crisis features institutions that borrowed too much based on optimistic assumptions about future returns. The housing bubble exemplified this dynamic perfectly: everyone from homebuyers to investment banks increased leverage based on the belief that housing prices only go up.

But measuring leverage isn't as simple as dividing debt by equity. Modern financial systems feature byzantine chains of borrowing and lending, with leverage embedded in derivatives positions, securities lending arrangements, and off-balance-sheet vehicles. It's like trying to measure someone's alcohol consumption when they're drinking cocktails mixed with other cocktails (and some of the ingredients aren't even labeled).

Interconnectedness transforms individual leverage problems into systemic ones. When institutions are highly connected through counterparty relationships and common asset holdings, one failure can cascade through the system like dominoes. Network analysis helps map these connections, but here's the challenge: networks change rapidly during crises as institutions scramble to reduce exposures.

**Interconnectedness paradox & learning objective**

Financial institutions often look *highly correlated*, yet correlation is **symmetric** and cannot tell us *who transmits* shocks to *whom*.  
Our goal is to estimate **market-implied directional connectedness** using the Diebold–Yilmaz framework based on the **generalized forecast-error variance decomposition (GFEVD)** of a reduced-form VAR.

**We will:**
1. Build a VAR on daily log-returns of major U.S. banks.
2. Compute the **row-normalized** GFEVD matrix $\;\tilde{\Theta}^{(H)}=\{\tilde{\theta}_{ij}^{(H)}\}\;$ where $\tilde{\theta}_{ij}^{(H)}$ is the share of the $H$-day ahead forecast-error variance of bank $i$ **due to shocks in** bank $j$.
3. Summarize with:
   - **Total connectedness** $C^{(H)} = 100\cdot \dfrac{\sum_{i\neq j}\tilde{\theta}_{ij}^{(H)}}{N}$,
   - **Directional "to others"** $C_{j\to \bullet}^{(H)} = 100\cdot \sum_{i\neq j}\tilde{\theta}_{ij}^{(H)}$,
   - **Directional "from others"** $C_{\bullet\to j}^{(H)} = 100\cdot \sum_{i\neq j}\tilde{\theta}_{ji}^{(H)}$,
   - **Net** $C_{j}^{\text{net}} = C_{j\to \bullet}^{(H)} - C_{\bullet\to j}^{(H)}$.

**Interpretation:** these are **directional, order-invariant variance shares** (reduced-form), not structural causal effects.


```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import networkx as nx
from statsmodels.tsa.api import VAR

plt.rcParams["figure.figsize"] = (10.5, 5.0)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

# Universe: six major U.S. banks (systemically relevant, liquid, long history)
tickers = ["JPM", "BAC", "C", "WFC", "GS", "MS"]

# Design choices (students can experiment with these):
start_date = "2006-01-01"   # long sample includes 2007–2009 crisis
window_H   = 500            # rolling window length in trading days
stride_s   = 5              # stride (days) for rolling estimation (captures dynamics without excessive noise)
H_FEVD     = 10             # forecast horizon for variance decomposition (days)
seed_layout = 7             # fixed seed for network layout reproducibility



#prices = yf.download(tickers, start=start_date, progress=False, auto_adjust = True)["Close"].dropna()
prices = pd.read_csv("/content/prices_US_banks.csv", index_col = "Date", parse_dates = True)
logp = np.log(prices)
rets = logp.diff().dropna()   # daily log-returns in decimals
rets.columns.name = None

# Stress proxy: VIX (daily), standardized for overlay; aligned to rets by forward-fill on business days
#vix = yf.download("^VIX", start=start_date, progress=False, auto_adjust = True)["Close"].rename(columns = {"^VIX": "VIX"}).dropna()
#vix = vix.reindex(rets.index, method="ffill")
vix = pd.read_csv("/content/vix.csv", index_col = "Date", parse_dates = True)
vix_z = ((vix - vix.mean()) / vix.std()).rename(columns = {"VIX":"VIX_z"})
```

**Correlation is symmetric; connectedness is directional**

Correlation matrices answer **“how much do they move together?”** but not **“who transmits to whom?”**  
We visualize:
1. The **correlation heatmap** of returns (symmetric).
2. **Cumulative return indices** (rebased to 1) to see common and idiosyncratic trends.

These motivate the need for a **VAR + GFEVD** approach to uncover directional spillovers.


```python
# correlation heatmap & cumulative return indices

# Correlation heatmap
C = rets.corr()

fig, ax = plt.subplots(1, 2, figsize=(12, 4.8))

im = ax[0].imshow(C.values, cmap="viridis", vmin=0, vmax=1)
ax[0].set_xticks(range(len(tickers))); ax[0].set_xticklabels(tickers, rotation=45)
ax[0].set_yticks(range(len(tickers))); ax[0].set_yticklabels(tickers)
ax[0].set_title("Return Correlation (symmetric)")
plt.colorbar(im, ax=ax[0], fraction=0.046)

# Cumulative return indices (rebased to 1)
cum_idx = np.exp(rets.cumsum())            # equivalent: cumulative product in log space

for col in tickers:
    ax[1].plot(cum_idx.index, cum_idx[col], lw=1.2, label=col)
ax[1].set_title("Cumulative Return Indices (rebased)")
ax[1].set_xlabel("Date")


ax[1].legend(ncol=3, fontsize=9, frameon=False)
plt.tight_layout()
plt.show()

print(f"Mean pairwise correlation: {C.values[np.triu_indices(len(tickers),1)].mean():.2f}  |  Range: [{C.min().min():.2f}, {C.max().max():.2f}]")
```

**VAR(p) and generalized FEVD: objects we estimate**

Let $y_t \in \mathbb{R}^N$ be the vector of bank returns. We fit a reduced-form VAR($p$):
$$
y_t = \Phi_1 y_{t-1} + \cdots + \Phi_p y_{t-p} + u_t,\quad u_t \sim (0,\Sigma_u).
$$

MA($\infty$) representation:
$$
y_t = \sum_{h=0}^{\infty} \Psi_h u_{t-h}.
$$

**Generalized FEVD** (Pesaran–Shin, order-invariant) at horizon $H$:
$$
\theta_{ij}^{(H)}
= \frac{\sigma_{jj}^{-1}\sum_{h=0}^{H-1}\left(e_i^\top \Psi_h \Sigma_u e_j\right)^2}
{\sum_{h=0}^{H-1} e_i^\top \Psi_h \Sigma_u \Psi_h^\top e_i},
\qquad
\tilde{\theta}_{ij}^{(H)}=\frac{\theta_{ij}^{(H)}}{\sum_{k=1}^N \theta_{ik}^{(H)}}
$$

**Diebold–Yilmaz indices:**
- Total: $C^{(H)} = 100 \cdot \dfrac{\sum_{i\neq j}\tilde{\theta}_{ij}^{(H)}}{N}$  
- To-others (from $j$): $C_{j\to \bullet}^{(H)} = 100\cdot \sum_{i\neq j} \tilde{\theta}_{ij}^{(H)}$  
- From-others (to $j$): $C_{\bullet\to j}^{(H)} = 100\cdot \sum_{i\neq j} \tilde{\theta}_{ji}^{(H)}$  
- Net: $C_j^{\text{net}} = C_{j\to \bullet}^{(H)} - C_{\bullet\to j}^{(H)}$

**Row/column meaning:** Row $i$ aggregates **variance of $i$ due to others**; Column $j$ aggregates **variance exported by $j$ to others**.


```python
# Choose a fixed lag order p by AIC on the first window of length H
first_window = rets.iloc[:window_H]
order_results = VAR(first_window).select_order(maxlags=10)
p = int(order_results.aic)

print(f"Chosen VAR order p = {p} (by AIC on first {window_H} days)")
```

We now implement a **single function** that maps a return matrix on a window $W$ to:
1. The row-normalized GFEVD matrix $\tilde{\Theta}^{(H)}$,
2. Total connectedness $C^{(H)}$,
3. Directional indices ($\text{To}$, $\text{From}$) and $\text{Net}$ for each bank.

Sanity checks: each row of $\tilde{\Theta}^{(H)}$ sums to $1$; off-diagonal mass is the source of connectedness.


```python
# DY function and static full-sample table and bar chart

def dy_connectedness(window_returns, p, H):

    model = VAR(window_returns)
    res = model.fit(p)

    # Moving-average coefficients Ψ_h for h=0,...,H-1
    # res.ma_rep(q) returns array of shape (q+1, N, N) with Ψ_0 = I_N
    Psi = res.ma_rep(maxn=H-1)
    Sigma = res.sigma_u.values
    N = window_returns.shape[1]

    # Precompute σ_jj (diagonal of Σ) once
    sigma_jj = np.diag(Sigma)

    # Numerator and denominator in θ_ij^(H)
    num = np.zeros((N, N))
    den = np.zeros(N)

    for h in range(H):
        A = Psi[h] @ Sigma              # A = Ψ_h Σ
        B = Psi[h] @ Sigma @ Psi[h].T   # B = Ψ_h Σ Ψ_h'
        num += (A[:, :] ** 2) / sigma_jj  # broadcast on columns j for σ_jj^{-1}
        den += np.diag(B)

    theta = num / den[:, None]              # θ_ij^(H)
    theta_tilde = theta / theta.sum(axis=1, keepdims=True)  # row-normalize

    # Indices (in percent)
    np.fill_diagonal(theta_tilde, theta_tilde.diagonal())   # unchanged diagonal (for clarity)
    total_C = 100.0 * (theta_tilde.sum() - np.trace(theta_tilde)) / N

    col_sums = theta_tilde.sum(axis=0) - np.diag(theta_tilde)  # "to others" (by column, excluding diagonal)
    row_sums = theta_tilde.sum(axis=1) - np.diag(theta_tilde)  # "from others" (by row, excluding diagonal)

    to_others = pd.Series(100.0 * col_sums, index=window_returns.columns, name="To")
    from_others = pd.Series(100.0 * row_sums, index=window_returns.columns, name="From")
    net_measure = (to_others - from_others).rename("Net")

    theta_tilde_df = pd.DataFrame(theta_tilde, index=window_returns.columns, columns=window_returns.columns)
    return theta_tilde_df, float(total_C), to_others, from_others, net_measure

# Static (full-sample) connectedness for an initial, didactic snapshot
theta_full, total_full, to_full, from_full, net_full = dy_connectedness(rets, p=p, H=H_FEVD)

# Presentation: connectedness table and "To vs From" bars
fig, ax = plt.subplots(1, 2, figsize=(12, 4.8))

# Heatmap
im = ax[0].imshow(theta_full.values, cmap="magma")
ax[0].set_xticks(range(len(tickers))); ax[0].set_xticklabels(tickers, rotation=45)
ax[0].set_yticks(range(len(tickers))); ax[0].set_yticklabels(tickers)
ax[0].set_title(r"$\tilde{\Theta}^{(H)}$ (row-normalized GFEVD)")
plt.colorbar(im, ax=ax[0], fraction=0.046)

# Bars: To vs From per bank
x = np.arange(len(tickers))
ax[1].bar(x - 0.2, to_full.values, width=0.4, label="To others")
ax[1].bar(x + 0.2, from_full.values, width=0.4, label="From others")
ax[1].set_xticks(x); ax[1].set_xticklabels(tickers, rotation=0)
ax[1].set_title("Directional connectedness (%)")
ax[1].legend(frameon=False)

plt.tight_layout()
plt.show()

print(f"Static total connectedness C^(H) = {total_full:.1f}%  (H={H_FEVD} days)")
```

We estimate connectedness on **rolling windows** of length $H=\,$500 trading days with stride $s=\,$5 days.  
Each window yields $C_t^{(H)}$ (total), $C_{j\to\bullet,t}^{(H)}$ (to), $C_{\bullet\to j,t}^{(H)}$ (from), and $C_{j,t}^{\text{net}}$ (net).  
We **overlay** standardized VIX $Z_t$ and annotate notable events (Bear Stearns, Lehman, COVID, SVB) to contextualize—but **not** to claim causality.

**Units:** indices are in **percent of $H$-day forecast-error variance**.


```python
# Rolling estimation and total connectedness with stress overlay and events

dates = []
C_total = []
to_store = []
from_store = []
net_store = []
theta_store = []

# Set business day frequency, forward-fill missing data
rets = rets.asfreq('B').fillna(method='ffill')

# roll windows with stride s
for t0 in range(0, len(rets) - window_H, stride_s):
    W = rets.iloc[t0:t0 + window_H]
    theta_t, C_t, to_t, from_t, net_t = dy_connectedness(W, p=p, H=H_FEVD)

    dates.append(W.index[-1])
    C_total.append(C_t)
    to_store.append(to_t.values)
    from_store.append(from_t.values)
    net_store.append(net_t.values)
    theta_store.append(theta_t.values)

# assemble time series
dates = pd.to_datetime(dates)
C_total = pd.Series(C_total, index=dates, name="Total")
To_df    = pd.DataFrame(to_store,    index=dates, columns=tickers)
From_df  = pd.DataFrame(from_store,  index=dates, columns=tickers)
Net_df   = pd.DataFrame(net_store,   index=dates, columns=tickers)

# plotting with VIX overlay and event annotations
fig, ax1 = plt.subplots(figsize=(11.5, 4.6))
ax1.plot(C_total.index, C_total.rolling(3).mean(), lw=1.8, label=r"$C^{(H)}$ (smoothed)")
ax1.set_ylabel("Total connectedness (%)")
ax1.set_title(r"Rolling Total Connectedness $C^{(H)}$ and Stress Proxy (VIX)")

ax2 = ax1.twinx()
ax2.plot(vix_z.reindex(C_total.index), lw=1.0, alpha=0.7, label="VIX (z-score)")
ax2.set_ylabel("VIX (z)")

# event annotations (dates approximate, illustrative)
events = {
    "Bear Stearns": "2008-03-16",
    "Lehman": "2008-09-15",
    "COVID shock": "2020-03-16",
    "SVB": "2023-03-10",
}
for label, d in events.items():
    d = pd.to_datetime(d)
    if C_total.index[0] <= d <= C_total.index[-1]:
        ax1.axvline(d, ls="--", color="gray", alpha=0.5)
        ax1.text(d, ax1.get_ylim()[1]*0.93, label, rotation=90, va="top", ha="right", fontsize=8)

ax1.legend(loc="upper left", frameon=False)
ax2.legend(loc="upper right", frameon=False)
plt.tight_layout()
plt.show()

print(f"Windows estimated: {len(C_total)}  |  mean C^(H)={C_total.mean():.1f}%  |  H={window_H}, stride={stride_s}")
```


We zoom into **2007–2009** to see acceleration around the Global Financial Crisis.  
We then visualize the **directed, weighted network** at the **exact peak** of $C^{(H)}$ and compare it to a **calm** window.

- Edge $j\to i$ weight: $\tilde{\theta}_{ij}^{(H)}$ (off-diagonal only).  
- Node size: $(\text{To}_j + \text{From}_j)$ at that window.  
- Node color: $\text{Net}_j$ (red = transmitter, blue = receiver).


```python
# Select windows in the crisis interval based on their end date
mask_crisis = (C_total.index >= "2007-01-01") & (C_total.index <= "2009-12-31")
C_crisis = C_total[mask_crisis]

# Time series zoom
fig, ax = plt.subplots(figsize=(11.5, 4.2))
ax.plot(C_crisis.index, C_crisis.rolling(3).mean(), lw=1.8)
ax.set_title(r"Zoom: $C^{(H)}$ during 2007–2009 (smoothed)")
ax.set_ylabel("Total connectedness (%)")
for label, d in {"Bear": "2008-03-16", "Lehman": "2008-09-15", "TARP": "2008-10-03"}.items():
    d = pd.to_datetime(d)
    if C_crisis.index[0] <= d <= C_crisis.index[-1]:
        ax.axvline(d, ls="--", color="gray", alpha=0.5)
        ax.text(d, ax.get_ylim()[1]*0.95, label, rotation=90, va="top", ha="right", fontsize=8)
plt.tight_layout()
plt.show()

# Identify peak and calm windows globally
idx_peak = C_total.idxmax()
idx_calm = C_total.idxmin()

# Retrieve matrices and directional indices at those dates
# We stored theta_store aligned with 'dates'
peak_i = dates.get_loc(idx_peak)
calm_i = dates.get_loc(idx_calm)

theta_peak = pd.DataFrame(theta_store[peak_i], index=tickers, columns=tickers)
theta_calm = pd.DataFrame(theta_store[calm_i], index=tickers, columns=tickers)

to_peak, from_peak = To_df.loc[idx_peak], From_df.loc[idx_peak]
to_calm, from_calm = To_df.loc[idx_calm], From_df.loc[idx_calm]
net_peak = (to_peak - from_peak)
net_calm = (to_calm - from_calm)

def draw_network(theta_df, to_s, from_s, net_s, title):
    # build directed graph from off-diagonal θ̃; threshold tiny edges for readability
    G = nx.DiGraph()
    for j in tickers:
        G.add_node(j)

    weights = []
    for i in tickers:
        for j in tickers:
            if i == j:
                continue
            w = float(theta_df.loc[i, j])
            if w >= 0.10:  # show edges >= 5% spillover
                G.add_edge(j, i, weight=w)
                weights.append(w)

    pos = nx.spring_layout(G, seed=seed_layout)  # reproducible layout
    node_size = 500 + 20 * (to_s + from_s)       # size ∝ total degree (to+from)
    # color by net (red for positive/net transmitter, blue for negative)
    net_vals = net_s.values
    net_norm = (net_vals - net_vals.min()) / (net_vals.max() - net_vals.min() + 1e-12)
    node_colors = plt.cm.coolwarm(net_norm)

    plt.figure(figsize=(7.6, 6.2))
    nx.draw_networkx_nodes(G, pos, node_size=node_size.values, node_color=node_colors)
    nx.draw_networkx_labels(G, pos, font_size=10)
    widths = [3.0 * G[u][v]["weight"] for u, v in G.edges()]  # scale edge widths by weight
    nx.draw_networkx_edges(G, pos, width=widths, arrows=True, arrowsize=14, alpha=0.8)
    plt.title(title)
    plt.axis("off")
    plt.show()

draw_network(theta_peak, to_peak, from_peak, net_peak, f"Network at PEAK connectedness ({idx_peak.date()})")
draw_network(theta_calm, to_calm, from_calm, net_calm, f"Network at CALM connectedness ({idx_calm.date()})")

# Identify top transmitter at peak
top_transmitter = to_peak.idxmax()
print(f"Peak at {idx_peak.date()}  |  Top transmitter (Net/To highest): {top_transmitter}")
```

We compare **average spillover matrices** in low-stress vs high-stress regimes and track **directional roles** over time.

- Regimes: bottom vs top quartiles of the stress proxy (standardized VIX).
- Matrices: average $\tilde{\Theta}^{(H)}$ in each regime and their **difference** (crisis $-$ calm).
- Roles: time series of $C_{j\to\bullet}^{(H)}$, $C_{\bullet\to j}^{(H)}$, and $C_{j}^{\text{net}}$ identify persistent **transmitters** and **receivers**.


```python
# Map each window end-date to a stress value (VIX_z at that date)
stress_at_window = vix_z.reindex(dates)
q_low, q_high = stress_at_window.quantile([0.25, 0.75]).values.flatten()

low_idx  = stress_at_window.index[(stress_at_window <= q_low).squeeze().values]
high_idx = stress_at_window.index[(stress_at_window >= q_high).squeeze().values]

# Average θ̃ in each regime
theta_low_mean  = np.mean([theta_store[dates.get_loc(d)] for d in low_idx], axis=0)
theta_high_mean = np.mean([theta_store[dates.get_loc(d)] for d in high_idx], axis=0)

theta_low_mean  = pd.DataFrame(theta_low_mean,  index=tickers, columns=tickers)
theta_high_mean = pd.DataFrame(theta_high_mean, index=tickers, columns=tickers)
theta_diff = theta_high_mean - theta_low_mean  # crisis minus calm

fig, ax = plt.subplots(1, 3, figsize=(15, 4.8))

im0 = ax[0].imshow(theta_low_mean.values, cmap="magma")
ax[0].set_title(r"Calm: $\mathbb{E}[\tilde{\Theta}\,|\,\text{low stress}]$")
ax[0].set_xticks(range(len(tickers))); ax[0].set_xticklabels(tickers, rotation=45)
ax[0].set_yticks(range(len(tickers))); ax[0].set_yticklabels(tickers)
plt.colorbar(im0, ax=ax[0], fraction=0.046)

im1 = ax[1].imshow(theta_high_mean.values, cmap="magma")
ax[1].set_title(r"Crisis: $\mathbb{E}[\tilde{\Theta}\,|\,\text{high stress}]$")
ax[1].set_xticks(range(len(tickers))); ax[1].set_xticklabels(tickers, rotation=45)
ax[1].set_yticks(range(len(tickers))); ax[1].set_yticklabels(tickers)
plt.colorbar(im1, ax=ax[1], fraction=0.046)

# Difference heatmap: use a centered colormap to highlight increases/decreases
v = np.max(np.abs(theta_diff.values))
im2 = ax[2].imshow(theta_diff.values, cmap="bwr", vmin=-v, vmax=v)
ax[2].set_title("Difference (crisis − calm)")
ax[2].set_xticks(range(len(tickers))); ax[2].set_xticklabels(tickers, rotation=45)
ax[2].set_yticks(range(len(tickers))); ax[2].set_yticklabels(tickers)
plt.colorbar(im2, ax=ax[2], fraction=0.046)

plt.tight_layout()
plt.show()

# Directional and net roles over time
fig, axes = plt.subplots(3, 1, figsize=(11.5, 8.8), sharex=True)

# To-others
for col in tickers:
    axes[0].plot(To_df.index, To_df[col].rolling(3).mean(), lw=1.2, alpha=0.9, label=col)
axes[0].set_title(r"Directional 'To others' $C_{j\to\bullet}^{(H)}$ (%)")
axes[0].legend(ncol=3, fontsize=8, frameon=False)

# From-others
for col in tickers:
    axes[1].plot(From_df.index, From_df[col].rolling(3).mean(), lw=1.2, alpha=0.9)
axes[1].set_title(r"Directional 'From others' $C_{\bullet\to j}^{(H)}$ (%)")

# Net (To − From)
for col in tickers:
    axes[2].plot(Net_df.index, Net_df[col].rolling(3).mean(), lw=1.4, alpha=0.95)
axes[2].axhline(0, ls="--", color="k", alpha=0.6)
axes[2].set_title(r"Net connectedness $C_j^{\text{net}}$ (%)")
axes[2].set_xlabel("Date")

plt.tight_layout()
plt.show()

# "Interpret this number" pedagogical moment at the peak
jstar = to_peak.idxmax()                                  # exporter (column)
i_star = theta_peak[jstar].drop(index=jstar).idxmax()      # receiver (row), off-diagonal
value  = theta_peak.loc[i_star, jstar]
print(f"At the peak, exporter {jstar} explains {value:.3f} of {i_star}'s H-step FE variance.")
```

We check **design robustness** and a **bounded early-warning association**:

- Robustness: compare $C^{(H)}$ across $(H_{\text{FEVD}}, H)$ choices and report correlations.
- Predictive check: regress **future 10-day realized volatility** of the equal-weighted bank index on **lagged** $C^{(H)}$ (scatter + OLS line).  
This illustrates *predictive content*, not structural causality.

Finally, we **export** tidy outputs for replication (indices and regime-average matrices).


```python

# Robustness: vary H_FEVD and window_H, compute correlations with baseline C_total
specs = [("H10_W500", H_FEVD, window_H),
         ("H5_W500", 5, window_H),
         ("H20_W500", 20, window_H),
         ("H10_W400", H_FEVD, 400),
         ("H10_W600", H_FEVD, 600)]

C_specs = {"BASE": C_total}

for name, Hh, Hw in specs:
    dates_s, series = [], []
    for t0 in range(0, len(rets) - Hw, stride_s):
        W = rets.iloc[t0:t0 + Hw]
        _, C_t, _, _, _ = dy_connectedness(W, p=p, H=Hh)
        dates_s.append(W.index[-1])
        series.append(C_t)
    C_specs[name] = pd.Series(series, index=pd.to_datetime(dates_s)).rename(name)

# Align and compute correlations with BASE
C_aligned = pd.concat(C_specs.values(), axis=1).dropna()
C_aligned.columns = list(C_specs.keys())
corrs = C_aligned.corr()["BASE"].drop("BASE")

# Predictive association: future 10-day realized vol of equal-weighted index vs lagged C^(H)
eq_index = rets.mean(axis=1)                         # equal-weighted daily return
realized_vol_10 = (eq_index.rolling(10).std() * np.sqrt(252)).shift(-10)  # future vol proxy
C_for_pred = C_total.reindex(realized_vol_10.index)

# align and drop NaN
pred_df = pd.concat([C_for_pred.shift(1).rename("C_lag1"), realized_vol_10.rename("fut_vol10")], axis=1).dropna()

# simple OLS: fut_vol10 = a + b * C_lag1
X = np.vstack([np.ones(len(pred_df)), pred_df["C_lag1"].values]).T
beta = np.linalg.lstsq(X, pred_df["fut_vol10"].values, rcond=None)[0]
fit_line = X @ beta
SS_res = np.sum((pred_df["fut_vol10"].values - fit_line)**2)
SS_tot = np.sum((pred_df["fut_vol10"].values - pred_df["fut_vol10"].mean())**2)
R2 = 1 - SS_res/SS_tot

# scatter + fitted line
fig, ax = plt.subplots(figsize=(6.8, 5.2))
ax.scatter(pred_df["C_lag1"], pred_df["fut_vol10"], s=10, alpha=0.5)
# draw line on the observed x-range
xg = np.linspace(pred_df["C_lag1"].min(), pred_df["C_lag1"].max(), 100)
yg = beta[0] + beta[1] * xg
ax.plot(xg, yg, lw=2)
ax.set_xlabel(r"Lagged total connectedness $C^{(H)}_{t-1}$ (%)")
ax.set_ylabel("Future 10-day realized vol (annualized)")
ax.set_title("Predictive association (illustrative, reduced-form)")
plt.tight_layout()
plt.show()

print(f"Robustness corr with BASE: {corrs.round(2).to_dict()}  |  OLS slope={beta[1]:.3f}, R^2={R2:.2f}")
```

The example above took us from **symmetric co-movement** to **directional transmission**. The correlation heatmap and cumulative indices showed that large U.S. banks move together, but correlation cannot tell us *who sends* risk to *whom*.

Using a reduced-form VAR with order $p=1$ on daily log-returns and the **generalized FEVD**, we estimated the row-normalized matrix $\tilde{\Theta}^{(H)}$ at horizon $H=10$, where $\tilde{\theta}_{ij}^{(H)}$ is the share of bank $i$’s $H$-day forecast-error variance due to shocks in bank $j$. From $\tilde{\Theta}^{(H)}$ we built the Diebold–Yilmaz indices: the **total connectedness** $C^{(H)}=100\cdot \frac{\sum_{i\neq j}\tilde\theta_{ij}^{(H)}}{N}$, **directional “to others”** $C_{j\to\bullet}^{(H)}=100\sum_{i\neq j}\tilde\theta_{ij}^{(H)}$, **directional “from others”** $C_{\bullet\to j}^{(H)}=100\sum_{i\neq j}\tilde\theta_{ji}^{(H)}$, and **net** $C_j^{\text{net}}=C_{j\to\bullet}^{(H)}-C_{\bullet\to j}^{(H)}$.

The static snapshot delivered a large off-diagonal mass (total $C^{(H)}\approx 74\%$), which is expected for a tightly coupled U.S. banking sector: most of the $H$-day uncertainty of any one bank is explained by *others*. Rolling this estimator on $500$-day windows (stride $=5$) produced a time series $C_t^{(H)}$ that **spikes in systemic episodes**—a clear run-up around the 2008 crisis and a pronounced peak during the March–June 2020 COVID liquidity shock—while remaining elevated but cyclical in quieter periods.

At the **exact peak** (2020-06-30), the **network snapshot** (edges drawn only when $\tilde\theta_{ij}^{(H)}\ge 10\%$) revealed concentrated transmission paths and identified **JPM** as the leading **exporter** (top of $C_{j\to\bullet}^{(H)}$ and positive $C_j^{\text{net}}$); our off-diagonal reading $\tilde\theta_{i^*,j^*}^{(H)}$ provided a concrete “interpret this number” moment: *“exporter $j^*$ explains a fraction of $i^*$’s $H$-day forecast variance.”*

Comparing **calm vs crisis** $\tilde{\Theta}^{(H)}$ matrices, the **difference heatmap** (crisis $-$ calm) showed broad intensification of off-diagonal shares—evidence that **amplification** occurs via cross-institution channels in stress. Directional panels $C_{j\to\bullet}^{(H)}$, $C_{\bullet\to j}^{(H)}$, and $C_j^{\text{net}}$ traced **role dynamics** over time (transmitters vs receivers), reinforcing that systemic *roles* are **state-dependent**.

Sensitivity checks showed near-unity correlations across $(H,H_{\text{FEVD}})$ specifications, which is consistent with daily VAR(1) dynamics where impulse responses decay quickly; a simple lead–lag regression of future $10$-day realized volatility on lagged $C^{(H)}$ had near-zero slope and $R^2$, underscoring that connectedness is best read as a **contemporaneous state variable** rather than a stand-alone forecaster.

**What we proved** is that the DY framework cleanly separates *symmetric co-movement* from *directional spillovers*, quantifies **who** transmits risk **to whom**, and shows **how** those links intensify in crises—precisely the mechanism by which leverage and shocks become **systemic**.

**Why this matters:** for monitoring and policy, elevated $C^{(H)}$ and persistent positive $C_j^{\text{net}}$ argue for tighter **macro-prudential buffers**, **liquidity add-ons**, and **network-aware** stress tests. **Limits:** our measures are **order-invariant variance shares** from a reduced form (not structural causality), and they depend on design choices $(p,H,H_{\text{FEVD}},\text{window},\text{stride})$—choices we exposed and tested for robustness. In sum, the example fulfilled its purpose: it linked formal objects to transparent code, contrasted correlation with **connectedness**, and told a coherent crisis-transmission story with evidence the students could reproduce and interpret.


### **4.2 Regulatory Gaps and Arbitrage**

Financial innovation consistently outpaces regulation, creating gaps where risks can accumulate outside supervisory oversight. The shadow banking system exemplified this dynamic – institutions that looked like banks, acted like banks, but weren't regulated like banks. When the crisis hit, regulators discovered they'd been monitoring only half the financial system (was this oversight inevitable given innovation incentives?).

Regulatory arbitrage represents another persistent challenge. Global banks shift activities to less regulated jurisdictions, while financial innovations are often designed explicitly to circumvent existing rules. Credit default swaps before 2008 allowed institutions to take on credit risk without holding actual loans – it was regulatory arbitrage disguised as financial innovation.

The fundamental problem is that regulation is static while markets are dynamic. By the time regulators identify and address one gap, clever financial engineers have opened three more. It's like playing whack-a-mole with moles that learn from each other and adapt their strategies.

## **5. Computational Methods: Agent-Based Modeling for Crisis Scenario Analysis**

### **5.1 Foundations of Agent-Based Modeling**

Agent-based models represent a philosophical revolution in finance. Instead of assuming markets reach equilibrium through the actions of rational, representative agents, ABMs simulate the messy reality of heterogeneous actors following simple rules and learning from their mistakes. It's the difference between studying traffic flow by assuming all drivers are identical robots versus modeling actual drivers with different skill levels, risk preferences, and tendencies to road rage.

In financial ABMs, individual agents represent banks, traders, households, or other economic actors who make decisions based on local information and simple behavioral rules. Some follow fundamentals, others chase trends, still others act on noise or sentiment. The beauty of this approach is that complex system dynamics – crashes, bubbles, volatility clustering – emerge naturally from agent interactions without being explicitly programmed into the model.

Building a crisis-focused ABM requires several components. Agent types might include commercial banks that lend based on capital constraints, investment banks that trade and create securities, households that borrow and save, and a central bank that sets policy. Each follows behavioral rules that capture essential behaviors while remaining computationally tractable. But here's the challenge: which behaviors matter most for crisis dynamics (and how would you know without running the experiment)?

### **5.2 Crisis Simulation and Analysis**

The August 2007 Quant Meltdown provides a perfect case study for ABM analysis. An agent-based model would populate the market with different fund types – quant funds following similar factor strategies, traditional mutual funds, market makers, and noise traders. The quant funds' similar models create the crowding that characterized the actual event.

When we introduce a shock – perhaps one fund liquidating due to losses elsewhere – the simulation reveals how selling pressure propagates. Initial selling depresses prices in the fund's portfolio. Other quant funds with similar positions experience losses, triggering their risk management rules. More selling ensues, creating a feedback loop that accelerates as market makers withdraw liquidity and spreads widen.

ABMs excel at counterfactual analysis. What if circuit breakers had halted trading? What if position limits had prevented excessive crowding? What if central banks had intervened earlier? By rerunning simulations with different rules, we can evaluate which policies might have prevented or mitigated the crisis. It's like having a time machine that lets you test different historical scenarios.

The challenge lies in calibration. ABMs often have hundreds of parameters governing individual behaviors, making traditional estimation techniques impractical. Researchers use method of simulated moments, genetic algorithms, and machine learning to find parameter combinations that generate realistic market dynamics. But validation remains tricky: how do you verify a model's ability to predict unprecedented events?

## **6. Macro-Prudential Policy Design**

### **6.1 Macro-Prudential Tools**

Macro-prudential policy represents a fundamental shift from protecting individual institutions to safeguarding the entire system. It's the difference between ensuring each car has good brakes versus managing traffic flow to prevent multi-vehicle accidents. The insight is that systemically stable outcomes require more than individually sound institutions (why might this be counterintuitive to traditional regulators?).

Countercyclical capital buffers exemplify this approach. Banks must build extra capital during credit booms that can be drawn down during busts. This leans against banking's natural procyclicality – the tendency to lend freely when risks appear low and tighten credit when the economy most needs it. It's like requiring people to save more when times are good so they can spend when times are bad.

Sectoral tools provide more targeted interventions. Loan-to-value limits for mortgages cool housing markets without affecting broader credit conditions. Risk weights that increase for specific asset classes make speculation more expensive without prohibiting it. These scalpel-like approaches avoid the sledgehammer effects of economy-wide interest rate changes.

Liquidity requirements ensure institutions can survive temporary funding disruptions. The Liquidity Coverage Ratio requires banks to hold enough liquid assets to survive a 30-day crisis. The Net Stable Funding Ratio addresses longer-term maturity mismatches. These prevent the liquidity spirals that characterized previous crises, though they also encourage gaming as institutions optimize ratios rather than genuine liquidity.

### **6.2 Policy Transmission and Effectiveness**

Macro-prudential policy effectiveness depends on transmission mechanisms that remain poorly understood. When authorities raise capital buffer requirements, banks face unpalatable choices: raise expensive equity capital, reduce dividend payments, or shrink their balance sheets. Each response affects the economy differently, making policy calibration extremely challenging.

International spillovers complicate matters further. Tightening domestic credit conditions might simply push borrowing offshore, with domestic borrowers accessing foreign banks or bond markets. Property developers constrained by local LTV limits might seek foreign buyers not subject to these restrictions. It's like trying to control water flow by building dams while leaving all the tributaries unregulated.

The interaction with monetary policy creates additional complexity. Loose monetary policy intended to stimulate growth might fuel asset bubbles that macro-prudential measures seek to constrain. Conversely, tight macro-prudential policy might offset monetary stimulus. Coordinating these policies requires institutional frameworks that most countries are still developing (and why might this coordination be particularly difficult politically?).

## **7. Financial Stability Monitoring Using Deep Learning Anomaly Detection**

### **7.1 Deep Learning Architectures for Anomaly Detection**

Deep learning transforms financial surveillance from pattern recognition to pattern creation. Autoencoders learn to compress market data into efficient representations and then reconstruct the original data. When reconstruction errors spike, something unusual is happening – possibly the early stages of a crisis, though possibly just noisy data on a slow news day.

Variational autoencoders (VAEs) extend this approach by learning probability distributions rather than point estimates. This provides uncertainty measures alongside anomaly scores – crucial for distinguishing genuine threats from statistical flukes. Think of VAEs as providing confidence intervals for crisis predictions, though the confidence itself may be misplaced.

Long Short-Term Memory networks excel at detecting unusual patterns in financial time series. These architectures maintain internal memory that captures temporal dependencies, identifying when current market behavior deviates from established patterns. But here's the catch: financial markets evolve continuously, so what constitutes "normal" behavior changes over time (and how fast should the model adapt without losing crisis sensitivity?).

Graph neural networks represent the cutting edge for analyzing financial networks. These models operate directly on network structures, learning representations that capture both individual institution characteristics and their interconnections. GNNs can identify anomalous changes in network topology – perhaps an institution suddenly increasing its connectedness or synchronized stress signals across a cluster of institutions.

### **7.2 Implementation and Challenges**

Implementing deep learning for financial surveillance faces the eternal challenge of rare events. Crises are rare by definition, creating severely imbalanced datasets that confuse most machine learning algorithms. Semi-supervised approaches train primarily on normal data, treating deviations as potential anomalies. One-class learning techniques define a boundary around normal behavior, flagging anything outside as suspicious.

The interpretability problem looms large in regulatory contexts. Regulators need to understand why a model flags certain conditions as anomalous – "the neural network thinks something's wrong" won't suffice when decisions affect millions of people. Attention mechanisms, SHAP values, and layer-wise relevance propagation provide partial solutions, but deep learning models remain fundamentally opaque.

Feature engineering requires careful balance between domain expertise and algorithmic discovery. Including market microstructure measures (bid-ask spreads, order book imbalances) alongside price data helps distinguish between fundamental stress and temporary disruptions. Text data from news and earnings calls provides context that numerical data misses. But too much human involvement defeats the purpose of using machines to find patterns humans can't see.

The integration challenge involves balancing innovation with reliability. False positives that trigger unnecessary interventions can be as damaging as false negatives that miss genuine risks. Hybrid systems combining deep learning with traditional methods often provide the best balance, though they sacrifice some of the elegance that makes pure machine learning approaches attractive.

## **8. Conclusion**

Crisis management and prevention represents financial engineering at its most consequential. The tools and frameworks we've explored – from robust statistical methods to agent-based models to deep learning surveillance systems – represent remarkable advances in our ability to understand and respond to financial instability. Yet humility remains essential.

Every crisis teaches us something new about how financial systems fail. The savings and loan crisis of the 1980s taught us about interest rate risk and moral hazard. The 1987 stock market crash highlighted the dangers of portfolio insurance. The dot-com bubble revealed how irrational exuberance can persist longer than rational observers expect. The 2008 crisis demonstrated the systemic dangers of interconnectedness and excessive leverage. Each lesson led to new regulations, better risk management, and more sophisticated models.

But here's the uncomfortable truth: the next crisis will likely involve factors we haven't fully considered, interactions we haven't modeled, and behaviors we haven't observed. This doesn't invalidate the methods we've studied – it makes them more important. Robust statistical techniques that perform well under model uncertainty, agent-based models that can simulate unprecedented scenarios, and anomaly detection systems that identify patterns in vast data streams all improve our crisis preparedness, even if they can't guarantee crisis prevention.

The challenge for financial engineers is developing tools that account for human psychology while maintaining quantitative rigor. Markets are driven by human emotions – fear, greed, overconfidence, herding – yet our models often assume rational behavior. Agent-based approaches offer partial solutions by incorporating heterogeneous behaviors and bounded rationality. Behavioral finance insights inform better risk models. But ultimately, crisis management requires understanding that markets are social systems, not physical ones, and human behavior under stress can be remarkably unpredictable (why might this be both the greatest challenge and the source of our field's continued relevance?).
